```python
variables_calidad = [
    'Escolaridad',
    'Estado Civil',
    'Ancestro Racial',
    'Orientación Sexual',
    'Identidad de Género'
]

categorias_no_disponibles = [
    'No sabe / No informa',
    'No había sido implementada',
    'Sin información',
    'No aplica'
]

resumen_calidad = []

for variable in variables_calidad:

    total_registros = len(df)

    registros_no_disponibles = df[variable].isin(
        categorias_no_disponibles
    ).sum()

    porcentaje_no_disponible = round(
        (registros_no_disponibles / total_registros) * 100,
        2
    )

    porcentaje_util = round(
        100 - porcentaje_no_disponible,
        2
    )

    resumen_calidad.append([
        variable,
        porcentaje_util,
        porcentaje_no_disponible
    ])

# IMPORTANTE: esto va FUERA del for
tabla_calidad = pd.DataFrame(
    resumen_calidad,
    columns=[
        'Variable',
        'Información útil (%)',
        'Información no disponible (%)'
    ]
)

tabla_calidad

tabla_calidad
```

```python
#Crear la variable Jornada
mapeo_jornada = {
'00:00 a 02:59': 'Madrugada',
'03:00 a 05:59': 'Madrugada',

'06:00 a 08:59': 'Mañana',
'09:00 a 11:59': 'Mañana',

'12:00 a 14:59': 'Tarde',
'15:00 a 17:59': 'Tarde',

'18:00 a 20:59': 'Noche',
'21:00 a 23:59': 'Noche',

'Sin información': 'Sin información'
}

df['Jornada'] = df[
'Rango_Hora_Limpio'
].map(mapeo_jornada)
```

**Transformación 4. Normalización de textos**

```python
#Paso 1. Identificar columnas de texto
columnas_texto = df.select_dtypes(
include='object'
).columns

print("Columnas de texto encontradas:")
print(columnas_texto)
```

```python
# Paso 2. Eliminar espacios innecesarios

for col in columnas_texto:
    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
    )

print("Normalización completada")
```

**Transformación 5. Estandarización de nombres de variables**

```python
# Código para normalizar nombres de columnas
import unicodedata

def limpiar_nombre_columna(col):
    col = unicodedata.normalize('NFKD', col)
    col = col.encode('ascii', 'ignore').decode('utf-8')
    col = col.replace(' ', '_')
    return col

df.columns = [
    limpiar_nombre_columna(col)
    for col in df.columns
]

df.columns
```
