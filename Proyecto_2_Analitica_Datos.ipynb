**Análisis Multivariado**

```python
#Función para calcular V de Cramér

import pandas as pd
import numpy as np

from scipy.stats import chi2_contingency

def cramers_v(x, y):

    tabla = pd.crosstab(x, y)

    chi2 = chi2_contingency(tabla)[0]

    n = tabla.sum().sum()

    r, k = tabla.shape

    return np.sqrt(chi2 / (n * (min(r-1, k-1))))
```

```python
#Sexo vs Estado Civil
cv_estado = cramers_v(
    df["Sexo de la victima"],
    df["Estado Civil"]
)

print("V de Cramér:", round(cv_estado,4))
```

```python
#Sexo vs Escolaridad
cv_escolaridad = cramers_v(
    df["Sexo de la victima"],
    df["Escolaridad"]
)

print("V de Cramér:", round(cv_escolaridad,4))
```

```python
#Sexo vs Zona del Hecho
cv_zona = cramers_v(
    df["Sexo de la victima"],
    df["Zona del Hecho"]
)

print("V de Cramér:", round(cv_zona,4))
```

```python
resultado = pd.DataFrame({

    "Relación":[
        "Sexo - Estado Civil",
        "Sexo - Escolaridad",
        "Sexo - Zona"
    ],

    "V_Cramer":[
        cv_estado,
        cv_escolaridad,
        cv_zona
    ]
})

resultado
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(
    resultado["Relación"],
    resultado["V_Cramer"],
    color="teal"
)

plt.title("Intensidad de asociación entre variables")
plt.ylabel("V de Cramér")

plt.ylim(0,1)

plt.show()
```

**8.3 Desarrollo del Objetivo Específico 3**

```python
# 8.3 ANÁLISIS ESPACIO-TEMPORAL
# MATRIZ DE TASAS POR DEPARTAMENTO Y AÑO

matriz_tasas = pd.pivot_table(
    df_tasas,
    values="Tasa_100k",
    index="Departamento_del_hecho_DANE",
    columns="Ano_del_hecho"
)

matriz_tasas.round(2)
```

```python
# 8.3 ANÁLISIS ESPACIO-TEMPORAL
# MAPA DE CALOR DE TASAS POR DEPARTAMENTO Y AÑO

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(14,12))

sns.heatmap(
    matriz_tasas,
    cmap="YlOrRd",
    linewidths=0.3
)

plt.title(
    "Distribución espacio-temporal de las tasas de mortalidad\n"
    "por suicidio en Colombia (2015-2024)"
)

plt.xlabel("Año")
plt.ylabel("Departamento")

plt.tight_layout()

plt.show()
```

```python
# 8.3 ANÁLISIS ESPACIO-TEMPORAL
# PROMEDIO Y VARIABILIDAD DE LAS TASAS POR DEPARTAMENTO

zonas_criticas = (
    df_tasas
    .groupby("Departamento_del_hecho_DANE")
    .agg(
        Tasa_Promedio=("Tasa_100k","mean"),
        Desviacion=("Tasa_100k","std")
    )
    .reset_index()
    .sort_values(
        "Tasa_Promedio",
        ascending=False
    )
)

zonas_criticas.head(15)
```

```python
# 8.3.3 IDENTIFICACIÓN DE ZONAS CRÍTICAS
# TASA PROMEDIO Y ESTABILIDAD TERRITORIAL

import matplotlib.pyplot as plt

top15 = zonas_criticas.head(15)

plt.figure(figsize=(10,6))

plt.scatter(
    top15["Desviacion"],
    top15["Tasa_Promedio"],
    s=120,
    color="darkred"
)

for _, row in top15.iterrows():
    plt.text(
        row["Desviacion"] + 0.05,
        row["Tasa_Promedio"],
        row["Departamento_del_hecho_DANE"],
        fontsize=9
    )

plt.title(
    "Relación entre tasa promedio y variabilidad\n"
    "departamental de la mortalidad por suicidio"
)

plt.xlabel("Desviación estándar")
plt.ylabel("Tasa promedio por 100.000 habitantes")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

**8.3.4 EVOLUCIÓN ESPACIO-TEMPORAL**

```python
# VARIACIÓN ENTRE EL PRIMER Y EL ÚLTIMO AÑO DISPONIBLE

evolucion_departamentos = []

for depto in matriz_tasas.index:

    serie = matriz_tasas.loc[depto].dropna()

    if len(serie) >= 2:

        primer_anio = serie.iloc[0]
        ultimo_anio = serie.iloc[-1]

        variacion = (
            (ultimo_anio - primer_anio)
            / primer_anio
        ) * 100

        evolucion_departamentos.append([
            depto,
            primer_anio,
            ultimo_anio,
            variacion
        ])

import pandas as pd

evolucion_departamentos = pd.DataFrame(
    evolucion_departamentos,
    columns=[
        "Departamento",
        "Tasa_Inicial",
        "Tasa_Final",
        "Variacion_%"
    ]
)

evolucion_departamentos = (
    evolucion_departamentos
    .sort_values(
        "Variacion_%",
        ascending=False
    )
)

evolucion_departamentos.round(2).head(15)
```

```python
# 8.3.4 EVOLUCIÓN ESPACIO-TEMPORAL
# DEPARTAMENTOS CON MAYOR CRECIMIENTO RELATIVO

import matplotlib.pyplot as plt

top_crecimiento = (
    evolucion_departamentos
    .head(15)
)

plt.figure(figsize=(12,6))

bars = plt.barh(
    top_crecimiento["Departamento"],
    top_crecimiento["Variacion_%"],
    color="steelblue"
)

for bar in bars:
    plt.text(
        bar.get_width() + 5,
        bar.get_y() + bar.get_height()/2,
        f"{bar.get_width():.0f}%",
        va="center"
    )

plt.title(
    "Departamentos con mayor crecimiento relativo\n"
    "de la tasa de mortalidad por suicidio (2015-2024)"
)

plt.xlabel("Variación porcentual (%)")
plt.ylabel("Departamento")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.show()
```

```python
# 8.3.5 IDENTIFICACIÓN DE PERFILES TERRITORIALES DE RIESGO
# INTEGRACIÓN DE INTENSIDAD Y CRECIMIENTO

perfiles_riesgo = zonas_criticas.merge(
    evolucion_departamentos[
        ["Departamento", "Variacion_%"]
    ],
    left_on="Departamento_del_hecho_DANE",
    right_on="Departamento"
)

perfiles_riesgo[
    [
        "Departamento_del_hecho_DANE",
        "Tasa_Promedio",
        "Desviacion",
        "Variacion_%"
    ]
].sort_values(
    "Tasa_Promedio",
    ascending=False
).head(15)
```

```python
# 8.3.5 IDENTIFICACIÓN DE PERFILES TERRITORIALES DE RIESGO
# MATRIZ INTENSIDAD VS CRECIMIENTO

import matplotlib.pyplot as plt

plt.figure(figsize=(11,7))

plt.scatter(
    perfiles_riesgo["Variacion_%"],
    perfiles_riesgo["Tasa_Promedio"],
    s=100,
    color="darkred"
)

for _, row in perfiles_riesgo.iterrows():
    plt.text(
        row["Variacion_%"] + 2,
        row["Tasa_Promedio"],
        row["Departamento_del_hecho_DANE"],
        fontsize=8
    )

plt.axhline(
    perfiles_riesgo["Tasa_Promedio"].mean(),
    color="gray",
    linestyle="--"
)

plt.axvline(
    perfiles_riesgo["Variacion_%"].mean(),
    color="gray",
    linestyle="--"
)

plt.title(
    "Perfiles territoriales de riesgo:\n"
    "intensidad y crecimiento de la mortalidad por suicidio"
)

plt.xlabel("Variación porcentual (%)")
plt.ylabel("Tasa promedio por 100.000 habitantes")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```
